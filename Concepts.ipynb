```python
def process_status(code):
    match code:
        case 200:
            return "Success"
        case 400:
            return "Failure"
        case _:
            return "DEfault"

print(process_status(200))
print(process_status(400))
print(process_status(100))
```

```python
def process_commands(command):
    match command:
        case ["load",filename]:
            return f"loading {filename}"
        case ["move",direction,steps]:
            return f"moveing from {direction} using {steps}"
        case ["draw",*shapes]:
            return f"drawing shapes {shapes}"
        case _:
            return "default"

print(process_commands(["load","details.csv"]))
print(process_commands(["move","Right","cursor"]))
print(process_commands(["draw","Circle","square","Triangle"]))
```

```python
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
        
def analyze_point(pt):
        match pt:
            case Point(x=0, y=0):
                return "At the origin (0, 0)"
            case Point(x=0, y=y_val):
                return f"On the Y-axis at y={y_val}"
            case Point(x=x_val, y=0):
                return f"On the X-axis at x={x_val}"
            case Point(x=x_val, y=y_val):
                return f"Point at coordinates ({x_val}, {y_val})"
            case _:
                return "Not a valid Point"    
print(analyze_point(Point(0, 0)))
print(analyze_point(Point(0, 1)))
print(analyze_point(Point(1, 0)))
print(analyze_point(Point(1, 1)))
```

```python
def process_response(response):
    match response:
        case {"status": 200, "data": [first_item, *rest]}:
            return f"Success! First item is {first_item}, with {len(rest)} more items."
        case {"status": error_code, "error": {"message": msg}} :
            return f"Error {error_code}: {msg}"
        case _:
            return "Malformed response"

print(process_response({"status": 200, "data": ["apple", "banana", "cherry"]}))
# Output: Success! First item is apple, with 2 more items.

print(process_response({"status": 404, "error": {"message": "Not Found"}}))
# Output: Error 404: Not Found
```

```python
def calculate_total(tax, discount, subtotal):
    return subtotal + tax - discount

# Unpacking a list/tuple for positional arguments
numbers_tuple = (5.0, 10.0, 100.0)
print(calculate_total(*numbers_tuple))  # Equivalent to calculate_total(5.0, 10.0, 100.0)
# Output: 95.0

# Unpacking a dictionary for keyword arguments
params_dict = {"subtotal": 200.0, "tax": 10.0, "discount": 20.0}
print(calculate_total(**params_dict))  # Equivalent to calculate_total(subtotal=200.0, tax=10.0, discount=20.0)
# Output: 190.0
```

```python
import time

def timing_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)  # Forwarding arguments safely
        end_time = time.time()
        print(f"Function '{func.__name__}' took {end_time - start_time:.4f} seconds to run.")
        return result
    return wrapper

@timing_decorator
def heavy_calculation(n):
    return sum(i**2 for i in range(n))

heavy_calculation(1000000)
# Output: Function 'heavy_calculation' took 0.05xx seconds to run. (approximate)
```

```python
class Logger:
    def log(self, message, level="INFO", *tags, **metadata):
        print(f"[{level}] {message} | Tags: {tags} | Meta: {metadata}")

class AdvancedLogger(Logger):
    def log(self, message, level="INFO", *tags, **metadata):
        print("--- Log entry processing started ---")
        # Pass everything straight up to the parent class method
        super().log(message, level, *tags, **metadata)

logger = AdvancedLogger()
logger.log("System rebooted", "WARNING", "system", "crit", user="admin", ip="192.168.1.10")
# Output:
# --- Log entry processing started ---
# [WARNING] System rebooted | Tags: ('system', 'crit') | Meta: {'user': 'admin', 'ip': '192.168.1.10'}
```

```python
#1. Maintaining Encapsulated State (Private Variables)
def make_bank_account(initial_balance):
    balance = initial_balance  # "Private" variable

    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
        return balance

    def withdraw(amount):
        nonlocal balance
        if 0 < amount <= balance:
            balance -= amount
            return balance
        return "Insufficient funds"

    # Returning multiple closures bundled together as a dictionary
    return {"deposit": deposit, "withdraw": withdraw}

account = make_bank_account(100)
print(account["deposit"](50))   # Output: 150
print(account["withdraw"](30))  # Output: 120
# print(balance)  # Error! 'balance' is hidden inside the closure scope.
```

```python
def repeat(num_times):
    def decorator_repeat(func):
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator_repeat

@repeat(num_times=3)
def greet():
    print("Python!")

greet()
# Output:
# Python!
# Python!
# Python!
```

```python
def memoize(func):
    cache = {}  # Stored in the closure's enclosing scope

    def wrapper(n):
        if n not in cache:
            print(f"Calculating result for {n}...")
            cache[n] = func(n)
        return cache[n]

    return wrapper

@memoize
def expensive_fibonacci_like(n):
    if n < 2:
        return n
    return n * 2

print(expensive_fibonacci_like(5))  # Calculates and caches
print(expensive_fibonacci_like(5))  # Instantly returns from cache without recalculating
# Output:
# Calculating result for 5...
# 10
# 10
```

```python
#Decorators
import time
from functools import wraps
 
def timed(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        print(f"{func.__name__} took {time.time()-start:.4f}s")
        return result
    return wrapper
 
@timed
def slow_add(a, b):
    time.sleep(0.1)
    return a + b
 
slow_add(2, 3)
```

```python
#Authentication / Access Control Guard
def require_admin(func):
    def wrapper(user, *args, **kwargs):
        if user.get("role") != "admin":
            return "Access Denied: Admins only!"
        return func(user, *args, **kwargs)
    return wrapper

@require_admin
def delete_database(user):
    return "Database deleted successfully."

print(delete_database({"name": "Alice", "role": "user"}))   # Output: Access Denied: Admins only!
print(delete_database({"name": "Bob", "role": "admin"}))    # Output: Database deleted successfully.
```

```python
#Automatic Input Validation / Type Checking
def validate_positive_numbers(func):
    def wrapper(a, b):
        if a < 0 or b < 0:
            raise ValueError("Arguments must be positive numbers!")
        return func(a, b)
    return wrapper

@validate_positive_numbers
def multiply(a, b):
    return a * b

print(multiply(4, 5))   # Output: 20
print(multiply(-2, 5)) # Raises ValueError: Arguments must be positive numbers!
```

```python
#Automatic Retry Mechanism (Fault Tolerance)
import time

def retry(max_attempts=3, delay=1):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"Attempt {attempt} failed: {e}. Retrying in {delay}s...")
                    time.sleep(delay)
            raise Exception("All retry attempts failed.")
        return wrapper
    return decorator

@retry(max_attempts=2, delay=0.1)
def unstable_api_call():
    print("Connecting to server...")
    raise ConnectionError("Server timeout")

 unstable_api_call() will try twice before finally crashing
```

```python
#Function Output Caching (Memoization)
def cache_results(func):
    cache = {}
    def wrapper(n):
        if n not in cache:
            print(f"Computing fresh result for {n}...")
            cache[n] = func(n)
        else:
            print(f"Fetching from cache for {n}...")
        return cache[n]
    return wrapper

@cache_results
def expensive_square(n):
    return n * n

print(expensive_square(4))  # Computes fresh result
print(expensive_square(4))  # Fetches instantly from cache
# Output:
# Computing fresh result for 4...
# 16
# Fetching from cache for 4...
# 16
```

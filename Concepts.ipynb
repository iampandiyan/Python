```python
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
        
def analyze_point(pt):
        match pt:
            case Point(x=0, y=0):
                return "At the origin (0, 0)"
            case Point(x=0, y=y_val):
                return f"On the Y-axis at y={y_val}"
            case Point(x=x_val, y=0):
                return f"On the X-axis at x={x_val}"
            case Point(x=x_val, y=y_val):
                return f"Point at coordinates ({x_val}, {y_val})"
            case _:
                return "Not a valid Point"    
print(analyze_point(Point(0, 0)))
print(analyze_point(Point(0, 1)))
print(analyze_point(Point(1, 0)))
print(analyze_point(Point(1, 1)))
```

```python
def process_response(response):
    match response:
        case {"status": 200, "data": [first_item, *rest]}:
            return f"Success! First item is {first_item}, with {len(rest)} more items."
        case {"status": error_code, "error": {"message": msg}} :
            return f"Error {error_code}: {msg}"
        case _:
            return "Malformed response"

print(process_response({"status": 200, "data": ["apple", "banana", "cherry"]}))
# Output: Success! First item is apple, with 2 more items.

print(process_response({"status": 404, "error": {"message": "Not Found"}}))
# Output: Error 404: Not Found
```

```python
import time

def timing_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)  # Forwarding arguments safely
        end_time = time.time()
        print(f"Function '{func.__name__}' took {end_time - start_time:.4f} seconds to run.")
        return result
    return wrapper

@timing_decorator
def heavy_calculation(n):
    return sum(i**2 for i in range(n))

heavy_calculation(1000000)
# Output: Function 'heavy_calculation' took 0.05xx seconds to run. (approximate)
```

```python
class Logger:
    def log(self, message, level="INFO", *tags, **metadata):
        print(f"[{level}] {message} | Tags: {tags} | Meta: {metadata}")

class AdvancedLogger(Logger):
    def log(self, message, level="INFO", *tags, **metadata):
        print("--- Log entry processing started ---")
        # Pass everything straight up to the parent class method
        super().log(message, level, *tags, **metadata)

logger = AdvancedLogger()
logger.log("System rebooted", "WARNING", "system", "crit", user="admin", ip="192.168.1.10")
# Output:
# --- Log entry processing started ---
# [WARNING] System rebooted | Tags: ('system', 'crit') | Meta: {'user': 'admin', 'ip': '192.168.1.10'}
```

```python
#1. Maintaining Encapsulated State (Private Variables)
def make_bank_account(initial_balance):
    balance = initial_balance  # "Private" variable

    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
        return balance

    def withdraw(amount):
        nonlocal balance
        if 0 < amount <= balance:
            balance -= amount
            return balance
        return "Insufficient funds"

    # Returning multiple closures bundled together as a dictionary
    return {"deposit": deposit, "withdraw": withdraw}

account = make_bank_account(100)
print(account["deposit"](50))   # Output: 150
print(account["withdraw"](30))  # Output: 120
# print(balance)  # Error! 'balance' is hidden inside the closure scope.
```

```python
def repeat(num_times):
    def decorator_repeat(func):
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator_repeat

@repeat(num_times=3)
def greet():
    print("Python!")

greet()
# Output:
# Python!
# Python!
# Python!
```

```python
def memoize(func):
    cache = {}  # Stored in the closure's enclosing scope

    def wrapper(n):
        if n not in cache:
            print(f"Calculating result for {n}...")
            cache[n] = func(n)
        return cache[n]

    return wrapper

@memoize
def expensive_fibonacci_like(n):
    if n < 2:
        return n
    return n * 2

print(expensive_fibonacci_like(5))  # Calculates and caches
print(expensive_fibonacci_like(5))  # Instantly returns from cache without recalculating
# Output:
# Calculating result for 5...
# 10
# 10
```

```python
#Decorators
import time
from functools import wraps
 
def timed(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        print(f"{func.__name__} took {time.time()-start:.4f}s")
        return result
    return wrapper
 
@timed
def slow_add(a, b):
    time.sleep(0.1)
    return a + b
 
slow_add(2, 3)
```

```python
#Authentication / Access Control Guard
def require_admin(func):
    def wrapper(user, *args, **kwargs):
        if user.get("role") != "admin":
            return "Access Denied: Admins only!"
        return func(user, *args, **kwargs)
    return wrapper

@require_admin
def delete_database(user):
    return "Database deleted successfully."

print(delete_database({"name": "Alice", "role": "user"}))   # Output: Access Denied: Admins only!
print(delete_database({"name": "Bob", "role": "admin"}))    # Output: Database deleted successfully.
```

```python
#Automatic Input Validation / Type Checking
def validate_positive_numbers(func):
    def wrapper(a, b):
        if a < 0 or b < 0:
            raise ValueError("Arguments must be positive numbers!")
        return func(a, b)
    return wrapper

@validate_positive_numbers
def multiply(a, b):
    return a * b

print(multiply(4, 5))   # Output: 20
print(multiply(-2, 5)) # Raises ValueError: Arguments must be positive numbers!
```

```python
#Automatic Retry Mechanism (Fault Tolerance)
import time

def retry(max_attempts=3, delay=1):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"Attempt {attempt} failed: {e}. Retrying in {delay}s...")
                    time.sleep(delay)
            raise Exception("All retry attempts failed.")
        return wrapper
    return decorator

@retry(max_attempts=2, delay=0.1)
def unstable_api_call():
    print("Connecting to server...")
    raise ConnectionError("Server timeout")

 unstable_api_call() will try twice before finally crashing
```

```python
#Function Output Caching (Memoization)
def cache_results(func):
    cache = {}
    def wrapper(n):
        if n not in cache:
            print(f"Computing fresh result for {n}...")
            cache[n] = func(n)
        else:
            print(f"Fetching from cache for {n}...")
        return cache[n]
    return wrapper

@cache_results
def expensive_square(n):
    return n * n

print(expensive_square(4))  # Computes fresh result
print(expensive_square(4))  # Fetches instantly from cache
# Output:
# Computing fresh result for 4...
# 16
# Fetching from cache for 4...
# 16
```

```python
#Resolving Conflicts with Explicit Parent Calls
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

class Colored:
    def __init__(self, color):
        self.color = color

class ColoredRectangle(Rectangle, Colored):
    def __init__(self, width, height, color):
        # Explicitly calling initializers from both parents
        Rectangle.__init__(self, width, height)
        Colored.__init__(self, color)

cr = ColoredRectangle(10, 5, "blue")
print(f"{cr.color} rectangle of size {cr.width}x{cr.height}")
# Output: blue rectangle of size 10x5
```

```python
#Comparison Operators (__eq__, __lt__)
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.price == other.price

    def __lt__(self, other):
        return self.price < other.price

p1 = Product("Laptop", 1000)
p2 = Product("Phone", 1000)
p3 = Product("Mouse", 50)

print(p1 == p2)  # Output: True (prices match)
print(p3 < p1)   # Output: True (50 < 1000)
```

```python
#Container and Length Emulation (__len__, __getitem__)
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

    def __getitem__(self, index):
        return self.members[index]

dev_team = Team(["Alice", "Bob", "Charlie"])

print(len(dev_team))    # Output: 3 (triggers __len__)
print(dev_team[1])      # Output: Bob (triggers __getitem__)
```

```python
#Making Objects Iterable (__iter__, __next__)
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        self.current -= 1
        return self.current + 1

for num in Countdown(3):
    print(num, end=" ")
# Output: 3 2 1
```

```python
#Context Managers (__enter__, __exit__)
import time

class TimerContext:
    def __enter__(self):
        self.start = time.time()
        print("--- Timer Started ---")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        elapsed = time.time() - self.start
        print(f"--- Timer Stopped: {elapsed:.4f}s elapsed ---")

with TimerContext():
    time.sleep(0.2)
# Output:
# --- Timer Started ---
# --- Timer Stopped: 0.20xxs elapsed ---
```

```python
#Type Conversion and Data Sanitization
class Profile:
    def __init__(self, email):
        self.email = email  # Triggers the setter on initialization too!

    @property
    def email(self):
        return self._email

    @email.setter
    def email(self, value):
        if "@" not in value:
            raise ValueError("Invalid email address")
        # Automatically clean input by lowercasing it
        self._email = value.strip().lower()

p = Profile("  USER@Example.COM  ")
print(p.email)  # Output: user@example.com
```

```python
#Managing Deprecated or Renamed Attributes
class Employee:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    @full_name.setter
    def full_name(self, name):
        # Splits a full name string back into first and last names
        parts = name.split(" ")
        self.first_name = parts[0]
        self.last_name = parts[1]

emp = Employee("John", "Doe")
print(emp.full_name)  # Output: John Doe
emp.full_name = "Jane Smith"
print(emp.first_name)  # Output: Jane
print(emp.last_name)   # Output: Smith
```

```python
#Enforcing Interface Compliance (Plug-and-Play Architecture)
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    @abstractmethod
    def process_payment(self, amount: float) -> bool:
        pass

class StripeProcessor(PaymentProcessor):
    def process_payment(self, amount: float) -> bool:
        print(f"Charged ${amount} via Stripe.")
        return True

class PayPalProcessor(PaymentProcessor):
    def process_payment(self, amount: float) -> bool:
        print(f"Charged ${amount} via PayPal.")
        return True

def checkout(processor: PaymentProcessor, amount: float):
    # Accepts any processor as long as it implements PaymentProcessor
    processor.process_payment(amount)

checkout(StripeProcessor(), 99.99)
checkout(PayPalProcessor(), 49.50)
```

```python
#Providing Shared Code alongside Abstract Requirements
from abc import ABC, abstractmethod

class Database(ABC):
    @abstractmethod
    def connect(self):
        pass  # Subclass must define how it connects

    def log_action(self, action):
        # Shared method available to all subclasses out-of-the-box
        print(f"[LOG]: Performing action -> {action}")

class PostgreSQLDB(Database):
    def connect(self):
        print("Connected to PostgreSQL database.")

db = PostgreSQLDB()
db.connect()         # Custom implementation
db.log_action("SELECT")  # Inherited shared method
```

```python
#Alternative Constructors (@classmethod)
class Date:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    @classmethod
    def from_dash_string(cls, date_str):
        # Parses a string like "2026-07-26"
        year, month, day = map(int, date_str.split("-"))
        return cls(year, month, day)

today = Date.from_dash_string("2026-07-26")
print(today.year)  # Output: 2026
```
